# 🏥 MVP - Análise de Eficiência e Custos das Internações Hospitalares no RS (SIH/DATASUS)

## 📌 Contexto do Projeto
Este projeto consiste na construção de um pipeline de dados de ponta a ponta na plataforma **Databricks**, utilizando a **Arquitetura Medallion** (Camadas Bronze, Silver e Gold). O objetivo é processar microdados públicos de internações do Sistema Único de Saúde (SUS) para gerar visões analíticas sobre custos e eficiência operacional.

---

## ❓ Perguntas de Negócio

1. **Gestão de Custos (CIDs & Procedimentos):**
   * *Quais são as patologias (CIDs) e procedimentos que concentram o maior custo total e maior valor médio por internação?*

2. **Gestão Operacional (Leitos & Permanência):**
   * *Qual é o tempo médio de permanência (dias em leito) por especialidade médica e faixa etária?*

3. **Qualidade & Desfechos Clínicos:**
   * *Qual é a taxa de mortalidade hospitalar cruzada por caráter de atendimento (Eletivo vs. Urgência/Emergência) e faixa etária?*

4. **Disparidade Regional:**
   * *Existe variação significativa no custo médio por internação entre os diferentes municípios e/ou regiões do RS?*
